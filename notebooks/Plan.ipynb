**Course:** Partial Differential Equations

**Project topic:** Analytical and Numerical Methods for the Black--Scholes Equation in European Option Pricing

---

Partial differential equations are mathematical models involving an unknown function of several variables and its partial derivatives. They are used to describe many phenomena in which a quantity changes with respect to time and space, for example heat conduction, diffusion, wave propagation, fluid flow or electrostatic potential.

In this project, we study a partial differential equation arising in financial mathematics: the Black--Scholes equation. This equation describes the price of a European option, which is a financial contract whose payoff is determined at a fixed maturity time.

The unknown function in our model is the option value

$$
V = V(S,t),
$$

where:

* $S$ is the price of the underlying asset,
* $t$ is time,
* $V(S,t)$ is the value of the option at time $t$ when the asset price is $S$.

Since the option value depends on both the asset price and time, the pricing problem naturally leads to a partial differential equation.

---

# 2. Main Idea of the Project

The Black--Scholes equation has the form

$$
V_t + \frac{1}{2}\sigma^2 S^2 V_{SS} + rS V_S - rV = 0.
$$

This is a second-order linear parabolic partial differential equation. It is not identical to the classical heat equation, but after a suitable change of variables it can be transformed into a heat-type equation.

This connection is the main mathematical motivation of the project. We want to show that the Black--Scholes model is not only a financial formula, but also an important example of a parabolic PDE.

The project has two main parts:

1. **Analytical part**: derivation of the Black--Scholes PDE, transformation to the heat equation, and presentation of the exact Black--Scholes formula.

2. **Numerical part**: finite difference approximation of the Black--Scholes PDE, comparison with the exact formula, convergence analysis, and a practical comparison with BTCUSDT option market prices.

The role of the heat equation is analytical. We use it to show that the Black--Scholes equation belongs to the class of parabolic PDEs and can be connected with methods studied in the PDE course. The numerical method is then applied directly to the original Black--Scholes PDE.

# 4. Mathematical Background

## 4.1 Partial Differential Equations

A partial differential equation is an equation involving an unknown function of several independent variables and its partial derivatives.

In this project, the unknown function is

$$
V = V(S,t).
$$

The variable $S$ plays the role of a spatial variable, while $t$ is the time variable. Therefore, the derivatives appearing in the model are

$$
V_t = \frac{\partial V}{\partial t},
$$

$$
V_S = \frac{\partial V}{\partial S},
$$

and

$$
V_{SS} = \frac{\partial^2 V}{\partial S^2}.
$$

The term $V_t$ describes the change of the option price with respect to time. The term $V_S$ describes the sensitivity of the option price with respect to the underlying asset price. The second derivative $V_{SS}$ describes the curvature of the option price as a function of $S$.

The presence of the second-order derivative $V_{SS}$ is one of the reasons why the Black--Scholes equation belongs to the class of second-order PDEs.

---

## 4.2 Parabolic PDEs and the Heat Equation

One of the most important examples of a parabolic PDE is the heat equation

$$
u_t = \alpha^2 u_{xx},
$$

where $u(x,t)$ denotes the temperature at position $x$ and time $t$, and $\alpha^2$ is the thermal diffusivity coefficient.

The heat equation describes diffusion: heat spreads from regions of higher temperature to regions of lower temperature. Mathematically, this smoothing effect is represented by the second-order derivative $u_{xx}$.

The Black--Scholes equation has a different interpretation because it describes the price of a financial derivative, not temperature. However, from the mathematical point of view, it has a similar parabolic structure. After a suitable transformation, the Black--Scholes equation can be reduced to a heat equation.

This transformation is important because it connects option pricing with the theory of parabolic PDEs studied in the course.

# 5. Financial Background

## 5.1 European Options

An option is a financial contract that gives the holder the right, but not the obligation, to buy or sell an underlying asset at a fixed price.

The fixed price is called the strike price and is denoted by $K$.

A European option can be exercised only at the maturity time $T$.

There are two basic types of European options.

### European Call Option

A European call option gives the holder the right to buy the underlying asset at the strike price $K$ at maturity $T$.

Its payoff is

$$
\max(S_T - K,0),
$$

where $S_T$ is the price of the underlying asset at maturity.

### European Put Option

A European put option gives the holder the right to sell the underlying asset at the strike price $K$ at maturity $T$.

Its payoff is

$$
\max(K - S_T,0).
$$

The goal of option pricing is to determine the fair value of the option before maturity, that is, to compute

$$
V(S,t)
$$

for $0 \leq t < T$.

---



# 6. Assumptions of the Black--Scholes Model

The Black--Scholes model is based on the following assumptions:

1. The underlying asset price follows a geometric Brownian motion:

$$
dS_t = \mu S_t,dt + \sigma S_t,dW_t,
$$

where:

* $\mu$ is the expected rate of return of the asset,
* $\sigma > 0$ is the volatility,
* $W_t$ is a standard Brownian motion.

2. The risk-free interest rate $r$ is constant.

3. The volatility $\sigma$ is constant.

4. There are no transaction costs.

5. Trading is continuous.

6. The underlying asset pays no dividends.

7. The option is European, so it can be exercised only at maturity.

These assumptions are idealized, but they allow us to derive a closed mathematical model for the option price.

A key feature of the model is that the expected return $\mu$ of the underlying asset disappears from the final PDE. The final option price depends on the risk-free rate $r$, volatility $\sigma$, strike price $K$, time to maturity $T$, and current asset price $S$.

---


# 7. Analytical Part of the Project

The analytical part of the project consists of four main steps.

First, we derive the Black--Scholes PDE from the stochastic model of the underlying asset price. The asset price is assumed to satisfy

$$
dS_t = \mu S_t,dt + \sigma S_t,dW_t.
$$

Then, applying Ito's formula to the option value $V(S,t)$ and constructing a risk-free hedging portfolio leads to the Black--Scholes PDE

$$
V_t + \frac{1}{2}\sigma^2 S^2 V_{SS} + rS V_S - rV = 0.
$$

Second, we formulate terminal and boundary conditions for European call and put options. The terminal condition is given by the payoff at maturity.

For a European call option,

$$
V(S,T)=\max(S-K,0).
$$

For a European put option,

$$
V(S,T)=\max(K-S,0).
$$

Third, we transform the Black--Scholes PDE into a heat-type equation. This step shows that the equation is a parabolic PDE and connects the project with classical PDE theory.

Fourth, we present the exact Black--Scholes formulas for European call and put options. These exact formulas will be used as benchmarks for the numerical method.

---


# 8. Numerical Part of the Project

The numerical part of the project is based on solving the original Black--Scholes PDE using a finite difference method.

We do not solve the heat equation numerically as a separate problem. The heat equation is used in the analytical part to understand the structure of the Black--Scholes equation.

The main numerical steps are:

1. Implement the exact Black--Scholes formula in Python.

2. Solve the Black--Scholes PDE numerically using a finite difference method.

3. Compare the finite difference solution with the exact Black--Scholes formula.

4. Compute the approximation error.

5. Study how the error changes when the computational grid is refined.

6. Analyze the sensitivity of option prices to model parameters.

7. Compare Black--Scholes model prices with observed BTCUSDT option market prices.

The numerical validation of the finite difference method is done by comparing it with the exact Black--Scholes formula, not directly with market prices. Market prices are used later as a practical illustration of how the theoretical model behaves when confronted with real data.

# 9. Computational Setup

Unless stated otherwise, we use the following baseline parameters:

$$
K = 100,
$$

$$
T = 1,
$$

$$
r = 0.05,
$$

$$
\sigma = 0.20.
$$

Here:

* $K$ is the strike price,
* $T$ is the time to maturity,
* $r$ is the risk-free interest rate,
* $\sigma$ is the volatility of the underlying asset.

The asset price is considered on a finite interval

$$
S \in [0,S_{\max}],
$$

where $S_{\max}$ is chosen sufficiently large compared with $K$. In the numerical examples, we take

$$
S_{\max} = 200.
$$

This finite interval approximates the theoretical domain

$$
S \in [0,\infty).
$$

For the finite difference method, we introduce a grid in the asset price variable and in time:

$$
S_i = i\Delta S,
$$

$$
t_j = j\Delta t.
$$

The terminal payoff is imposed at time $T$, and the PDE is solved backwards in time from $T$ to $0$.

---




# 10. Exact Black--Scholes Formula as a Benchmark

The exact Black--Scholes formula for a European call option is

$$
C(S,t)=SN(d_1)-Ke^{-r(T-t)}N(d_2),
$$

where

$$
d_1 =
\frac{\ln(S/K)+\left(r+\frac{1}{2}\sigma^2\right)(T-t)}
{\sigma\sqrt{T-t}},
$$

and

$$
d_2=d_1-\sigma\sqrt{T-t}.
$$

The exact Black--Scholes formula for a European put option is

$$
P(S,t)=Ke^{-r(T-t)}N(-d_2)-SN(-d_1).
$$

Here, $N(\cdot)$ denotes the cumulative distribution function of the standard normal distribution.

These exact formulas are important because they provide benchmark solutions. After computing the numerical solution of the PDE, we compare it with the exact Black--Scholes price.

The main error measure is

$$
\text{Error}(S_i)=|V_{\text{num}}(S_i,0)-V_{\text{exact}}(S_i,0)|.
$$

We also use global error measures such as the maximum absolute error

$$
\max_i |V_{\text{num}}(S_i,0)-V_{\text{exact}}(S_i,0)|
$$

and the root mean squared error

$$
\sqrt{
\frac{1}{M}
\sum_{i=1}^{M}
\left(
V_{\text{num}}(S_i,0)-V_{\text{exact}}(S_i,0)
\right)^2
}.
$$

---




# 11. Finite Difference Approximation of the Black--Scholes PDE

The Black--Scholes PDE is

$$
V_t+\frac{1}{2}\sigma^2S^2V_{SS}+rSV_S-rV=0.
$$

For a European call option, the terminal condition is

$$
V(S,T)=\max(S-K,0).
$$

The boundary conditions are approximated by

$$
V(0,t)=0,
$$

and

$$
V(S_{\max},t)\approx S_{\max}-Ke^{-r(T-t)}.
$$

The finite difference method replaces derivatives by discrete approximations on a grid. Since the terminal condition is known at maturity $T$, the solution is computed backwards in time.

In this project, we use a finite difference method to compute

$$
V_{\text{num}}(S,0).
$$

Then we compare it with the exact price

$$
V_{\text{exact}}(S,0).
$$

The goal is to verify that the numerical PDE method gives values close to the analytical Black--Scholes formula.

---


# 12. Grid Convergence Analysis

A numerical method should become more accurate when the computational grid is refined.

To test this, we solve the Black--Scholes PDE using different numbers of grid points, for example

$$

$$

$$
(M,N)=(100,100),
$$

$$
(M,N)=(200,200),
$$

and

$$
(M,N)=(400,400).
$$

For each grid, we compute the numerical solution and compare it with the exact Black--Scholes formula.

The goal is to observe whether the error decreases when the grid becomes finer.

This part is important because it shows that the finite difference method is not only producing a graph, but also approximating the PDE solution in a systematic way.

# 13. Parameter Sensitivity Analysis

The Black--Scholes price depends on several parameters:

$$
S,\quad K,\quad T,\quad r,\quad \sigma.
$$

In this part, we analyze how the option price changes when one parameter is varied while the others are fixed.

We consider the following cases.

### Volatility

We compute option prices for different values of volatility, for example

$$
\sigma = 0.10,\quad 0.20,\quad 0.40.
$$

A higher volatility usually increases the value of both call and put options, because greater uncertainty increases the probability of favorable outcomes for the option holder.

### Interest Rate

We compute option prices for different values of the risk-free interest rate, for example

$$
r = 0.01,\quad 0.05,\quad 0.10.
$$

For call options, a higher interest rate usually increases the option price, because the present value of the strike payment decreases.

### Strike Price

We compute option prices for different strike prices, for example

$$
K = 80,\quad 100,\quad 120.
$$

For call options, a higher strike price decreases the option value.

### Time to Maturity

We compute option prices for different maturities, for example

$$
T = 0.25,\quad 0.5,\quad 1,\quad 2.
$$

A longer time to maturity usually increases the value of an option because there is more time for the underlying asset price to move favorably.

The main mathematical part of the project validates the finite difference method by comparing it with the exact Black--Scholes formula.

This part should not be interpreted as a validation of the finite difference method itself. The finite difference method is validated against the exact Black--Scholes formula. The BTCUSDT comparison is used to show how the theoretical model behaves when compared with real option market data.

The comparison has the following structure:

$$
\text{market option price}
\quad \text{vs} \quad
\text{exact Black--Scholes price}
\quad \text{vs} \quad
\text{finite difference Black--Scholes PDE price}.
$$

The expected interpretation is:

* the exact Black--Scholes price and the finite difference price should be close to each other,
* the market price may differ from the Black--Scholes price,
* the difference between the market price and the model price can be caused by model assumptions, implied volatility, bid--ask spread, liquidity, market demand, and the specific behavior of crypto markets.

## 14.2 Historical Volatility Estimation for BTCUSDT

If implied volatility is not used, we can estimate the volatility from historical BTCUSDT prices.

Let

$$
S_0,S_1,\ldots,S_n
$$

be historical closing prices of BTCUSDT.

The logarithmic returns are defined by

$$
R_i=\ln\left(\frac{S_i}{S_{i-1}}\right),
\quad i=1,\ldots,n.
$$

The daily volatility is estimated as

$$
\sigma_{\text{daily}}=\operatorname{std}(R_1,\ldots,R_n).
$$

Assuming approximately 365 trading days in a year for the cryptocurrency market, the annualized volatility is

$$
\sigma_{\text{annual}}=\sigma_{\text{daily}}\sqrt{365}.
$$

This differs from the stock market convention, where one often uses approximately 252 trading days per year. Since BTCUSDT trades continuously, using 365 days is more natural for crypto data.

The estimated annualized volatility can then be used in the Black--Scholes formula and in the finite difference PDE solver.

## 14.3 Black--Scholes Price for the Selected BTCUSDT Option

After selecting one BTCUSDT option contract, we compute its theoretical Black--Scholes price.

For a call option, the exact Black--Scholes price is

$$
C_{\text{BS}}=S_0N(d_1)-Ke^{-rT}N(d_2),
$$

where

$$
d_1=
\frac{\ln(S_0/K)+\left(r+\frac{1}{2}\sigma^2\right)T}
{\sigma\sqrt{T}},
$$

and

$$
d_2=d_1-\sigma\sqrt{T}.
$$

For a put option, the exact Black--Scholes price is

$$
P_{\text{BS}}=Ke^{-rT}N(-d_2)-S_0N(-d_1).
$$

The same input parameters are also used in the finite difference solver. This gives the numerical PDE price

$$
V_{\text{FD}}(S_0,0).
$$

Then we compare the three values:

$$
V_{\text{market}},
$$

$$
V_{\text{BS}},
$$

and

$$
V_{\text{FD}}.
$$

## 14.4 Interpretation of the BTCUSDT Market Comparison

The BTCUSDT market comparison is not expected to match the Black--Scholes model perfectly.

The Black--Scholes model assumes constant volatility, continuous trading, no transaction costs, and a lognormal model for the underlying asset price. Real crypto option markets may violate these assumptions.

Possible reasons for differences between the market price and the Black--Scholes model price include:

1. The market uses implied volatility, which may be different from historical volatility.

2. Volatility is not constant in reality.

3. Crypto markets often have high volatility and large price jumps.

4. The bid--ask spread may be significant.

5. Liquidity may vary between option contracts.

6. Market participants may price risk differently from the simple Black--Scholes model.

7. The risk-free interest rate used in the model is only an approximation.

Therefore, the goal of this section is not to prove that Black--Scholes perfectly prices BTCUSDT options. The goal is to show how a theoretical PDE-based model can be compared with real option market prices and to discuss the limitations of the model.
